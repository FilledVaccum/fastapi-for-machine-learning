```python
# Doing this to make sure both notebook and terminal uses same pip location

import os
import sys
import subprocess
# Add environment's bin directory to PATH
env_bin = os.path.dirname(sys.executable)
current_path = os.environ['PATH']
if env_bin not in current_path:
    os.environ['PATH'] = f"{env_bin}:{current_path}"

# Verify fix
result = subprocess.run(['which', 'pip'], capture_output=True, text=True)
print("Fixed pip location:", result.stdout.strip())
```

```python
from fastapi import FastAPI
import json
```

### Loading data patient.json

```python
# loading data patient.json

def load_data():
    with open('patients.json', 'r') as f:
        data = json.load(f)

    return data
```

```python
app = FastAPI()

@app.get("/")
def hello():
    return { "message": "Patient Management System API" }

@app.get("/about")
def about():
    return { "messages" : "You are in about section broooo - This is fully functional API to manage patient records" }


@app.get("/view")
def view():
    data = load_data()
    return data
```
